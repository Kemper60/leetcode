```python
def reverseinteger(x):
    if abs(x) > 2**31:
        return(0)

    #log if the incoming number is positive or negative
    nchck = True if x < 0 else False 

    #convert this to the absolute value
    x = abs(x)

    #step 1 - convert the number to a list
    l = [i for i in str(x)]

    #step 2 - reverse the list
    l.reverse()

    #create an empty string
    s = ""

    #step 3 - convert back to a number
    for i in l:
        s = str(s) + str(i)

    #convert back to an integer
    f = int(s)

    #if it came in negative, make it negative again
    if nchck:
        f = f * -1

    #do another check to make sure this didn't take us out of bounds
    if abs(f) > 2**31 -1:
        return(0)

    #return the final number
    return (f)

    
```

```python
t = 120

print(reverseinteger(t))
```
