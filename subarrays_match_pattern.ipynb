```python
#accepted solution for https://leetcode.com/problems/number-of-subarrays-that-match-a-pattern-i/
#runtime beat 55% of others, memory beat 77% of others
class Solution:

    def countMatchingSubarrays(self, nums: list[int], pattern: list[int]) -> int:

        l = len(pattern) + 1 #get the length of the target pattern variable
        x = 0 #set an initial looping variable
        f = 0 #set a final count variable to return later
        ind = 0 #set another index variable

        #increment the looping variable
        while x < len(nums):

            #incrementally get a list of our target size
            chck = nums[x:x+l]

            #don't waste our time if the length of the check list is less than our target
            if len(chck) < l:
                break
            
            print(chck)
            
            #loop through the range and see if it matches our pattern
            for i in pattern:

                #increment index by 1
                ind = ind + 1

                #apply all the checks

                #if i is 1 then chck[ind] needs to be greater than chck[ind + 1] to fail
                if i == 1:
                    fail_val = chck[ind-1] >= chck[ind]

                #if i is 0 then chck[ind] needs to be not equal to chck[ind + 1] to fail
                elif i == 0:
                    fail_val = chck[ind-1] != chck[ind]

                #if i is -1 then chck[ind] needs to be less than to chck[ind + 1] to fail
                elif i == -1:
                    fail_val = chck[ind-1] <= chck[ind]

                #if we failed above then back out now
                if fail_val:
                    break

                #if we got to the end of the pattern list that means all the tests passed
                if ind == len(pattern):
                    f += 1
            
            ind = 0 #reset this
            x += 1  #increment x by 1

        #return the final counter
        return(f)
```

```python
sol = Solution()
sol.countMatchingSubarrays(nums=[376068870,376068870,376068870,412890845], pattern=[0, 1])
```
