```python
def lengthlongestsub(s):

    cnt = 1
    cnt_max = 1
    t = []

    if s == "":
        return(0)
    
    #loop through the input string
    for i in range(0, len(s)):

        print(i)

        #add our first value to our test list
        t.append(s[i])
        #print(i, s[i], t)

        #if the next character isn't in the test array capture it
        for x in range(i+1, len(s)):

            #if the next value isn't captured in our list
            if s[x] not in t:

                cnt += 1        #add to our counter
                t.append(s[x])  #capture it so it's no longer a new value
            else:

                #replace the max count with what we found
                cnt_max = max(cnt_max, cnt)

                #reset the count variable and the test list
                cnt = 1
                t = []

                #break this loop
                break

        #do one last check to replace the max count with what we found in the last run
        cnt_max = max(cnt_max, cnt)

    return cnt_max
```

```python
r = 'abcabcd'

lengthlongestsub(r)

```
