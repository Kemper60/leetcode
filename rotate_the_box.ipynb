```python
#accepted solution for leetcode problem https://leetcode.com/problems/rotating-the-box/
#faster than 80% of submissions

class Solution:

    def rotateTheBox(self, boxGrid: list[list[str]]) -> list[list[str]]:

        m = len(boxGrid) #number of lists
        n = len(boxGrid[0]) #length of list

        #create a transposed structure to hold this
        r = [list(row) for row in zip(*boxGrid)]

        # Reverse each row in the transpose grid to complete the 90° rotation
        for i in range(n):
            r[i].reverse()

        # Apply gravity to let stones fall to the lowest possible empty cell in each column
        for i in range(m):

            #set the lowest empty cell as the bottom minus one to start
            lowest_empty = n - 1

            # Process each cell in column `j` from bottom to top
            for x in range(n - 1, -1, -1):

                # Found a stone - let it fall to the lowest empty cell
                if r[x][i] == "#":
                    r[x][i] = "."
                    r[lowest_empty][i] = "#"
                    lowest_empty -= 1

                # Found an obstacle - reset `lowest_empty` to the row directly above it
                if r[x][i] == "*":
                    lowest_empty = x - 1

        return r
    
    
```

```python
sol = Solution()

b = [["#","#","*",".","*","."],["#","#","#","*",".","."],["#","#","#",".","#","."]]

sol.rotateTheBox(b)
```
