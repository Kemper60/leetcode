```python
#accepted solution for leetcode problem https://leetcode.com/problems/rotating-the-box/
#faster than 86% of submissions
#memory better than 74% of submissions

class Solution:

    def rotateTheBox(self, boxGrid: list[list[str]]) -> list[list[str]]:

        #make a new list to hold the items
        nl = []

        #loop through the input box list reversed. By inverting each list and working backwards
        #we can leverage the native zip() function later to complete the transposal
        for i in reversed(boxGrid):

            #set the minimum position we can drop a rock to
            min_pos = len(i)-1

            #loop backwards through each element of the list backwards so we can make the rock fall to the lowest point
            for x in range( len(i)-1, -1, -1):

                #found a rock, flip it to the minimum position we've found
                if i[x] == "#":
                    i[x] = "."
                    i[min_pos] = "#"
                    min_pos = min_pos-1

                #found a barrier, shift the minimum position up
                elif i[x] == "*":
                    min_pos = x-1

            #at the end of the inner loop append this to the new list
            nl.append(i)

        #return the transposed new list
        return([list(row) for row in zip(*nl)])
    
    
```

```python
sol = Solution()

b = [["#","#","*",".","*","."],
              ["#","#","#","*",".","."],
              ["#","#","#",".","#","."]]

sol.rotateTheBox(b)
```
