```python
#accepted solution for https://leetcode.com/problems/minimum-swaps-to-sort-by-digit-sum/
#beat 72% on speed, 75% on memory
class Solution:
    def minSwaps(self, nums: list[int]) -> int:
        n = len(nums)

        fl = [] #final 
        fx = 0

        #first convert everything to the sum of the digits
        for i in nums:

            #loop through each element as a string so we can parse it
            for x in str(i):

                #convert back to an integer and get the sum of the numbers
                fx = int(x) + fx

            fl.append(fx)
            fx = 0 #reset final x variable

        #get the target order inclusive of both the modified and original list
        indices = list(range(n))
        indices.sort(key=lambda i: (fl[i], nums[i]))

        #make an empty array of the correct size
        targetPos = [0] * n

        #map the indices in the targetPos
        #ex. - we want to place the index where the sorted value position would be
        for i, x in enumerate(indices):

            targetPos[x] = i

        #create a correctly sized array with all False values
        visited = [False] * n

        #set this to 0 to start
        cnt = 0


        #loop through the size of the list
        for i in range(n):

            x = 0

            #if we haven't visited this position yet, increment the counter and investigate more
            if not visited[i]:
                cnt += 1
                j = i

                #flip position j to true, then move j to the new position of what we swapped it for
                #keep going until we don't hit any more that need to be switched
                #total number of iterations is limited to the total number of input values
                while not visited[j]:
                    visited[j] = True
                    j = targetPos[j]
                    x += 1
                print(x)

        #return the number of input numbers less the times we found that something was already swapped
        return n - cnt
```
