```python
#accepted solution for problem https://leetcode.com/problems/simple-bank-system/
#solution was faster than 83.54% of submissions
class Bank:

    #initial call
    def __init__(self, balance: list[int]):

        #capture the accounts and balance information
        self.accounts = balance

        return(None)

    #transfer function - executes transfer and returns whether the transaction was valid
    def transfer(self, account1: int, account2: int, money: int) -> bool:

        #check that the account exists
        if account1 - 1 > len(self.accounts) or account2 - 1 > len(self.accounts):
            return(False) #return False because accounts are invalid
        
        #check that account 1 can cover their commitment
        elif self.accounts[account1-1] >= money:
            self.accounts[account2-1] = self.accounts[account2-1] + money #add the money to account 2
            self.accounts[account1-1] = self.accounts[account1-1] - money #remove money from account 1
            return(True) #let them know we did this
        else:

            #return false because they have insufficient funds
            return(False)

    #deposit function - executes deposit and returns whether the transaction was valid
    def deposit(self, account: int, money: int) -> bool:

        #check that the account exists
        if account - 1 > len(self.accounts):
            return(False) #return False because account is invalid

        else:
            #add the balance
            self.accounts[account - 1] = self.accounts[account - 1] + money

            #let them know we can do this one
            return(True)

    #withdraw function - executes withdrawal and returns whether the transaction was valid
    def withdraw(self, account: int, money: int) -> bool:

        #check that the account exists
        if account - 1 > len(self.accounts):
            return(False) #return False because account is invalid
                     
        elif self.accounts[account-1] >= money:
            self.accounts[account-1] = self.accounts[account-1] - money #remove money from account
            return(True) #let them know we did this

        else:

            #let them know they are too broke for this
            return(False)

# Your Bank object will be instantiated and called as such:
# obj = Bank(balance)
# param_1 = obj.transfer(account1,account2,money)
# param_2 = obj.deposit(account,money)
# param_3 = obj.withdraw(account,money)
```

```python
#sample solution

obj = Bank([10, 100, 20, 50, 30])


param_1 = obj.withdraw(3,10)
param_2 = obj.transfer(5,1,20)
param_3 = obj.deposit(5,20)
param_4 = obj.transfer(3,4,15)
param_5 = obj.withdraw(10,50)

print(param_1, param_2, param_3, param_4, param_5)
```
